```python
cost_price = float(input("Enter Cost Price: "))
selling_price = float(input("Enter Selling Price: "))

if selling_price > cost_price:
    profit = selling_price - cost_price
    profit_percentage = (profit / cost_price) * 100
    print(f"Profit: ₹{profit:.2f}")
    print(f"Profit Percentage: {profit_percentage:.2f}%")
elif selling_price < cost_price:
    loss = cost_price - selling_price
    loss_percentage = (loss / cost_price) * 100
    print(f"Loss: ₹{loss:.2f}")
    print(f"Loss Percentage: {loss_percentage:.2f}%")
else:
    print("No Profit, No Loss.")
```

```python
price = float(input("Enter product price: "))
gst_rate = float(input("Enter GST rate (in %): "))

gst_amount = price * gst_rate / 100
final_price = price + gst_amount

print(f"GST Amount: ₹{gst_amount:.2f}")
print(f"Final Price: ₹{final_price:.2f}")
```

```python
cost_price = float(input("Enter Cost Price: "))
selling_price = float(input("Enter Selling Price: "))
gst_rate = float(input("Enter GST Rate (%): "))
discount_rate = float(input("Enter Discount Rate (%): "))


if selling_price > cost_price:
    print(f"Profit: ₹{selling_price - cost_price:.2f}")
elif cost_price > selling_price:
    print(f"Loss: ₹{cost_price - selling_price:.2f}")
else:
    print("No Profit, No Loss")


discount_amount = selling_price * discount_rate / 100
price_after_discount = selling_price - discount_amount
print(f"Price after discount: ₹{price_after_discount:.2f}")
gst_amount = price_after_discount * gst_rate / 100
final_price = price_after_discount + gst_amount
print(f"Final price after GST: ₹{final_price:.2f}")
```
